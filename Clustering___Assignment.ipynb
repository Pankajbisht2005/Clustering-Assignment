Question 1: What is the difference between K-Means and Hierarchical Clustering? Provide a use case for each.

**K-Means Clustering**


- Definition

K-Means is a partition-based clustering algorithm that divides data into K predefined clusters.


- Number of Clusters

Must be specified in advance (K value).

- Working Principle

Minimizes within-cluster variance using centroid-based grouping.

- Speed & Scalability

Very fast and efficient for large datasets.

- Memory Requirement

Low memory usage.

- Cluster Shape

Works best for spherical and evenly sized clusters.

**Hierarchical clustering**

- Defination

Hierarchical clustering builds a tree-like structure (dendrogram) by merging or splitting clusters step-by-step.

- Number of Clusters

No need to predefine number of clusters. Clusters are chosen by cutting the dendrogram.

-Working Pinciple

Groups data based on distance and similarity, forming a hierarchy.
- Spead & Scalability

  Computationally expensive, slower for large datasets.
- Memory Requirement

  High memory usage because full distance matrix is needed.

Question 3: What are the core parameters of DBSCAN, and how do they influence the
clustering process?
- Introduction

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a density-based clustering algorithm that groups together data points that are closely packed, while marking low-density points as outliers (noise). Unlike K-Means, DBSCAN does not require the number of clusters to be specified in advance and can detect arbitrarily shaped clusters.

DBSCAN mainly depends on two core parameters:

1. ε (Epsilon)

2. MinPts (Minimum Points)

- Core Parameters of DBSCAN

1. Epsilon (ε)

Definition:
Epsilon (ε) defines the maximum distance between two data points for them to be considered neighbors.

Influence on Clustering:

Small ε value
→ Results in many small clusters and more noise points
→ Risk of splitting natural clusters.

Large ε value
→ Results in fewer but larger clusters
→ Risk of merging distinct clusters into one.

Impact:
ε directly controls the density threshold. A correct ε value ensures meaningful cluster formation.

2. MinPts (Minimum Points)

Definition:
MinPts is the minimum number of points required within the ε-neighborhood for a point to be considered a core point.

Influence on Clustering:

Small MinPts value
→ Produces many small clusters
→ Sensitive to noise and outliers.

Large MinPts value
→ Produces fewer clusters
→ May miss smaller but valid clusters.

Impact:
MinPts controls the minimum cluster density. Higher values create more robust and meaningful clusters.

Question 4: Why is feature scaling important when applying clustering algorithms like K-Means and DBSCAN?
- Introduction

Feature scaling is the process of standardizing the range of independent variables so that no single feature dominates the distance calculations. Since clustering algorithms like K-Means and DBSCAN rely heavily on distance metrics (such as Euclidean distance), feature scaling becomes critical for correct clustering results.

Without proper scaling, features with larger numeric ranges overpower smaller-scale features, leading to biased, incorrect, and misleading clusters.

- Importance of   Scaling features

1. Distance-Based Algorithms Are Sensitive to Magnitude

K-Means and DBSCAN use distance measures to form clusters. If one feature has values in thousands and another in fractions, the algorithm will prioritize the larger-valued feature, even if it is less important.

Example:

Customer dataset:

Age  18 to 60

Annual Income → ₹2,00,000 to ₹20,00,000

Without scaling:

 Income dominates

 Age becomes almost irrelevant

 Clusters become distorted

2. Prevents Biased Cluster Formation

Without feature scaling:

Clusters form mainly based on large-scale features

Smaller-scale but important features get ignored

Leads to wrong grouping

Scaling ensures equal contribution of all features, leading to balanced and meaningful clusters.

3. Improves Algorithm Performance and Stability

Feature scaling improves convergence speed in K-Means

Helps DBSCAN choose appropriate ε (epsilon) values

Produces more stable and reproducible clusters

Without scaling:

K-Means may converge slowly

DBSCAN struggles to find correct density thresholds

4. Essential for DBSCAN Parameter Selection

DBSCAN heavily depends on ε (epsilon) which is distance-based.

Without scaling:

Choosing ε becomes almost impossible

Wrong ε  either all noise or one giant cluster

With scaling:

ε values become interpretable and consistent

Question 5: What is the Elbow Method in K-Means clustering and how does it help determine the optimal number of clusters?
- Introduction

The Elbow Method is a technique used to determine the optimal number of clusters (k) in K-Means clustering. It works by analyzing the Within-Cluster Sum of Squares (WCSS), which measures the variance within each cluster.

>As the number of clusters increases, WCSS decreases because clusters become smaller and tighter.

> However, after a certain point, the rate of decrease slows down.

>The point where the curve bends (like an "elbow") indicates the optimal number of clusters, balancing accuracy and simplicity.

Steps in the Elbow Method
- Run K-Means clustering for different values of k (e.g., from 1 to 10).
- Calculate the WCSS for each k.
- Plot k (number of clusters) on the x-axis and WCSS on the y-axis.
- Identify the "elbow point" where the curve starts to flatten.
- This point suggests the best trade-off between minimizing WCSS and avoiding too many clusters.

```python
"""
Question 6: Generate synthetic data using make_blobs(n_samples=300, centers=4),
apply KMeans clustering, and visualize the results with cluster centers.
(Include your Python code and output in the code box below.)
"""
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate synthetic dataset
X, y = make_blobs(n_samples=300, centers=4, cluster_std=1.2, random_state=42)

# Apply KMeans clustering
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)
centers = kmeans.cluster_centers_

# Visualize the clusters and centroids
plt.scatter(X[:, 0], X[:, 1])
plt.scatter(centers[:, 0], centers[:, 1], marker='X', s=200)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering with Cluster Centers")
plt.show()
```

```python
"""
Question 7: Load the Wine dataset, apply StandardScaler , and then train a DBSCAN
model. Print the number of clusters found (excluding noise).
(Include your Python code and output in the code box below.)
"""
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import numpy as np

# Load Wine dataset
wine = load_wine()
X = wine.data

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train DBSCAN model
dbscan = DBSCAN(eps=0.8, min_samples=5)
labels = dbscan.fit_predict(X_scaled)

# Number of clusters (excluding noise)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

print("Number of clusters found (excluding noise):", n_clusters)
print("Unique labels:", set(labels))
```

```python
"""
Question 8: Generate moon-shaped synthetic data using
make_moons(n_samples=200, noise=0.1), apply DBSCAN, and highlight the outliers in
the plot.
(Include your Python code and output in the code box below.)
"""

import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import numpy as np

# Generate moon-shaped data
X, y = make_moons(n_samples=200, noise=0.1, random_state=42)

# Apply DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=5)
labels = dbscan.fit_predict(X)

# Identify outliers
outliers = labels == -1
clusters = labels != -1

# Plot clusters and outliers
plt.scatter(X[clusters, 0], X[clusters, 1], marker='o')
plt.scatter(X[outliers, 0], X[outliers, 1], marker='x')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("DBSCAN on make_moons Dataset (Outliers Highlighted)")
plt.show()

# Print cluster and outlier count
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_outliers = np.sum(outliers)

print("Number of clusters found:", n_clusters)
print("Number of outliers detected:", n_outliers)
```

```python
"""
Question 9: Load the Wine dataset, reduce it to 2D using PCA, then apply
Agglomerative Clustering and visualize the result in 2D with a scatter plot.
(Include your Python code and output in the code box below.)
"""
# Import required libraries
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler

#  Load the Wine dataset
wine = load_wine()
X = wine.data

#  Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

#  Reduce to 2D using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Apply Agglomerative Clustering
agglo = AgglomerativeClustering(n_clusters=3)  # Wine dataset has 3 classes
labels = agglo.fit_predict(X_pca)

# Visualize the clustering result
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='Set1', s=50, edgecolors='k')
plt.title('Agglomerative Clustering on Wine Dataset (2D PCA)')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.grid(True)
plt.show()
```

marketing team wants to segment customers based on their purchasing behavior to run
targeted promotions. The dataset contains customer demographics and their product

Describe your real-world data science workflow using clustering:

● Which clustering algorithm(s) would you use and why?

● How would you preprocess the data (missing values, scaling)?

● How would you determine the number of clusters?

● How would the marketing team benefit from your clustering analysis?
(Include your Python code and output in the code box below.)


-  Choice of Clustering Algorithm(s)
- K-Means: Efficient, scalable, and works well for large datasets like e-commerce customer records. Ideal for segmenting customers into distinct groups based on purchasing behavior.
- DBSCAN (optional): Useful if we suspect irregular or non-spherical clusters, or want to detect outliers (e.g., customers with unusual buying patterns).
- Agglomerative Clustering (optional): Helpful for hierarchical insights, e.g., seeing how small customer groups merge into larger segments.

2. Data Preprocessing
- Handle Missing Values:
- Fill missing demographics (e.g., age, income) with median values.
- Drop or impute missing purchase history depending on business rules.
- Feature Scaling:
- Apply StandardScaler or MinMaxScaler so that features like “Annual Income” and “Purchase Frequency” contribute equally.
- Feature Engineering:
- Create derived features such as average spend per category, recency of purchase, or loyalty score.

3. Determining the Number of Clusters
- Elbow Method: Plot WCSS vs. number of clusters to find the “elbow point.”
- Silhouette Score: Validate cluster quality by checking cohesion and separation.
- Business Context: Sometimes the “best” number of clusters is guided by marketing needs (e.g., 4–5 customer segments for targeted campaigns).

4. Benefits for the Marketing Team
- Targeted Promotions: Different clusters can receive personalized offers (e.g., discounts for budget shoppers, premium bundles for high spenders).
- Customer Retention: Identify loyal vs. at-risk customers and design retention strategies.
- Cross-Selling Opportunities: Spot customers who buy in one category but could be nudged toward another.
- Strategic Insights: Helps marketing understand customer diversity beyond simple averages.


```python
# example

# Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Step 1: Simulate a customer dataset (replace with real data in practice)
np.random.seed(42)
data = pd.DataFrame({
    'Age': np.random.randint(18, 65, 200),
    'Annual_Income': np.random.randint(20000, 120000, 200),
    'Purchase_Frequency': np.random.randint(1, 20, 200),
    'Avg_Spend': np.random.randint(100, 2000, 200)
})

# Preprocess (scaling)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(data)

# Determine optimal clusters using Elbow Method
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()

#  Apply KMeans with chosen k (say k=4)
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X_scaled)

#  Evaluate with Silhouette Score
score = silhouette_score(X_scaled, labels)
print("Silhouette Score:", score)

# Add cluster labels to dataset
data['Cluster'] = labels
print(data.head())
```
